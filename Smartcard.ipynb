# Data Preprocessing

### Handling Missing values

```python
data["Income"] = data["Income"].fillna(data["Income"].median())
```

```python
data.isnull().sum()
```

## Feature Engineering

```python
# Age
data["Age"] = 2026 -data["Year_Birth"]
```

```python
data.info()
```

```python
data["Dt_Customer"].max()
```

```python
# Customer Joining Date
data["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"], dayfirst=True)

reference_data = data["Dt_Customer"].max()

data["Customer_Tenure_Days"] = (reference_data - data["Dt_Customer"]).dt.days
```

```python
data.head()
```

```python
data["Total_Spending"] = data["MntWines"] + data["MntFruits"] + data["MntMeatProducts"] + data["MntFishProducts"] + data["MntSweetProducts"] + data["MntGoldProds"]
```

```python
data["Total_Children"] = data["Kidhome"] + data["Teenhome"]
```

```python
data.head()
```

```python
data["Education"] = df["Education"]
```

```python
# Education

data["Education"] = data["Education"].replace({
    "Basic":"Undergraduate","2n Cycle":"Undergraduate",
    "Graduation":"Graduate",
    "Master":"Postgraduate", "PhD":"Postgraduate"
})
```

```python
# Marital Status

data["Living_With"] = data["Marital_Status"].replace({
    "Married":"Partner","Together":"Partner",
    "Single":"Alone", "Divorced":"Alone",
    "Widow":"Alone", "Absurd":"Alone", "YOLO":"Alone"
})
```

```python
data["Education"].value_counts()
```

```python
data["Living_With"].value_counts()
```

## Visualize

```python
pca.explained_variance_ratio_
```

```python
# 3D 

pca = PCA(n_components=3)
X_pca = pca.fit_transform(X_scaled)
```

```python
pca.explained_variance_ratio_
```

```python
# plot 
fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111,projection="3d")

ax.scatter(X_pca[:,0], X_pca[:,1], X_pca[:,2])

ax.set_xlabel("pca1")
ax.set_ylabel("pca2")
ax.set_zlabel("pca2")
ax.set_title("3d projections")
```

# Analyze K value
## 1.Elbow Method

```python
from sklearn.cluster import KMeans
from kneed import KneeLocator

wcss = []
for k in range(1,11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit_predict(X_pca)
    wcss.append(kmeans.inertia_)
```

```python
knee = KneeLocator(range(1,11), wcss, curve="convex",direction="decreasing")
optimal_k = knee.elbow
```

```python
print(optimal_k)
```

```python
# plot 

plt.plot(range(1,11), wcss, marker="o")
plt.xlabel("K")
plt.ylabel("wcss")
```

## Silhouette Score

```python
from sklearn.metrics import silhouette_score

scores = []

for k in range(2,11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_pca)

    score = silhouette_score(X_pca, labels)
    scores.append(score)


# plot
plt.plot(range(2,11), scores, marker="o")
plt.xlabel("k")
plt.ylabel("Silhouette score")
```

```python
# combined plot

k_range = range(2,11)

fig, ax1 = plt.subplots(figsize=(8,6))

ax1.plot(k_range, wcss[:len(k_range)], marker="o", color="blue")
ax1.set_xlabel("K")
ax1.set_ylabel("wcss")

ax2 = ax1.twinx()
ax2.plot(k_range, scores[:len(k_range)], marker="x", color="red", linestyle="--")
ax2.set_ylabel("SS")
```

```python
print(len(k_range))
```

# Clustering

```python
# Kmeans

kmeans = KMeans(n_clusters=4,random_state=42)
labels_kmeans = kmeans.fit_predict(X_pca)
```

```python
fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111,projection="3d")

ax.scatter(X_pca[:,0], X_pca[:,1], X_pca[:,2], c=labels_kmeans)
```

```python
# Agglomerative Clustering

from sklearn.cluster import AgglomerativeClustering
```

```python
agg_clf = AgglomerativeClustering(n_clusters=4, linkage="ward")
labels_agg = agg_clf.fit_predict(X_pca)
```

```python
fig = plt.figure(figsize=(8,6))

ax = fig.add_subplot(111,projection="3d")

ax.scatter(X_pca[:,0], X_pca[:,1], X_pca[:,2], c=labels_agg)
```

# Characterization of cluster

```python

X["clusters"] = labels_agg
```

```python
pal = ["red","blue","yellow","green"]

sns.countplot(x=X["clusters"], palette=pal, hue=X["clusters"])
```
